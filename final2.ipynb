가로랑 세로로 두고 할일을 네칸으로 나누어서 위치시킨다.

- 가로축: 실행가능성
- 세로축: 비즈니스 임팩트

행렬은 행과 열로 구성되어있기 때문에, 가로에 해당하는 것을 가로축으로, 세로에 해당하는것을 세로축으로 하여금 도표로 그려볼 수 있다.

여러 과제들을 두가지 기준으로 평가해서 평면위에 점을 찍고, 기준마다 선을하나 그어서 사분면으로 나눈다.

기준선을 어디에 긋는가, 세로축을 어느 척도로 그리는가, 점의 크기와 색상에 무엇을 담는가에 대한 선택에따라 판단하는 사람의 기준이 달라질수 있다.

- 빠른 성과: 임팩트는 높음, 실행가능성은 높음
- 주요 과제: 임팩트는 높음, 실행가능성은 낮음
- 손쉬운 과제: 임팩트는 낮음, 실행가능성은 높음
- 재검토: 임팩트는 낮음, 실행가능성은 낮음

- 빠른성과: 비즈니스적으로 효과도 크고, 쉽게 할수있는 일 (가장먼저 추진해야하는 일)
- 주요과제: 비즈니스적으로 효과는 큰데, 실행하려면 문제가 있는 일 (문제를 해결하기 위한 계획을 먼저 세운다음에 추진)
- 손쉬운과제: 실행하는건 쉬운데, 비즈니스적으로 효과가 적은 것 (여력이 있을때 수행하는 일)
- 재검토: 비즈니스적으로도 효과가 적고, 하기도 어려운 일 (보류)

## 판단기준

사분면의 판정 기준이 있어야 한다.

임팩트가 기준선 이상인가 실행 가능성이 기준선 이상인가


## 같은 점, 다른기준

기준선은 평면을 4칸으로 나누는 두개의 선을 의미한다. 칸마다 가장 먼저 추진해야하는것, 문제를 해결하고 추진해야하는것들에 대한 대응이 정해져 있기 때문에 기준선이 조금만 움직여도 어떤 개선안은 대응이 완전 변할 수 있다.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False
```

```python
# 편익(Benefit)
# 어떤 선택, 투자, 정책 등을 통해서 얻게되는 긍정적인 효과 또는 경제적 이익

ideas = pd.DataFrame({
    '임팩트': [6000, 4500, 2940, 300, 800], # 연편익
    '실행가능성': [2.4, 4.2, 4.0, 4.6, 1.6 ]
}, index=[
    '야간 게이트 예약제',
    '선석 계획 공유 알림',
    '반입 예약 자동 확인',
    '안내 문구 정비',
    '통합 플랫폼 재구축'
])

impact_cut = 1000
feas_cut = 3.0

high_impact = ideas.loc['야간 게이트 예약제', '임팩트'] >= impact_cut
high_feas = ideas.loc['야간 게이트 예약제', '실행가능성'] >= feas_cut

print(high_impact, high_feas)
```

```python
def quadrant(impact, feas, impact_cut, feas_cut):
    return np.select(
        [
            (impact >= impact_cut) & (feas >= feas_cut),
            impact >= impact_cut,
            feas >= feas_cut,
        ],
        ['빠른성과', '주요 과제', '손쉬운 과제'], default='재검토'
    )
```

```python
ideas['사분면'] = quadrant(ideas['임팩트'], ideas['실행가능성'], impact_cut, feas_cut)
```
